```python
#--------------------------------------------

#          PYTHON CLASSES

#--------------------------------------------


#CLASS: FRAMEWORK TO CREATE OBJECTS, AND DO ACTIONS/APPLY METHODS TO THEM


#CONVENTION: start classes with capital letter
#          : start objects by small letter


#RMK) methods/functions defined inside a class hold only to the objects of that class

```

```python
#1) OBJECT CREATION WITH ATTRIBUTES
#standard syntax:

#class ClassName:
#    def __init__(self, attribute1, attribute2):
#        self.attribute1 = attribute1
#        self.attribute2 = attribute2


class Dog:
    def __init__(self, name):   #creates a new object with attribute name
        self.name = name       

class Dog1:
    def __init__(self, name, age):   #creates a new object with attributes name and age
        self.name = name
        self.age = age

mydog = Dog1("Jack", 10)    #the new object mydog with the given attributes
```

```python
#2) METHODS/ACTIONS IN THE CLASS to apply to the objects
#== functions inside the class, referring to the object as "self"


class Dog2:
    def __init__(self, name, age):    #1) initialisation of the object with attributes name and age
        self.name = name
        self.age = age

    def bark(self):     #2) method to apply to the object
                        #inside the class the object is called self
        print("Woof! My name is", self.name, "and I am", self.age, "years old.")


mydog2 = Dog2("Jack", 10)    #the new object mydog with the given attributes
mydog2.bark()                #call the method bark to the object mydog2
```

```python
#-------------------------

    #EXAMPLE: Rectangle class with methods to compute perimeter and area

#-------------------------

class Rectangle:
    def __init__(self, a, b):   #1) object initialisation with sides
        self.a = a
        self.b = b

    def perimeter(self):             #2) method to compute perimeter
        return 2*(self.a + self.b)

    def area(self):                  #2) method to compute area
        return self.a * self.b



rec = Rectangle(5, 10)         #new object rectangle created with attributes 5 and 10
print('Perimeter is', rec.perimeter())    #call method perimeter to the object rec
print('Area is', rec.area())              #call method area to the object rec
```

```python
#It is possible to define SUBCLASSES

class Square(Rectangle):
    def __init__(self, a):
        super().__init__(a, a)  #super() calls the constructor of the parent class (Rectangle)

    def modify_side(self, a1):
        super().modify_sides(a1, a1)  #super() calls the method of the parent class (Rectangle)

    #The following is completely useless, but it is possible to redefine the methods of the parent class

    def area(self):
        return super().area()  #super() calls the method of the parent class (Rectangle)

    def circle_inside_square(self):
        import numpy as np
        return np.pi * (self.a / 2) ** 2
```

```python
sq = Square(5)
print('Perimeter', sq.perimeter())
print('Area', sq.area())
print('Area of inscribed circle', sq.circle_inside_square())
```

```python
isinstance(sq, Square)  #check if sq is an instance of the class Square
```
